```python
# Generate realistic e-commerce funnel dataset (5000+ rows)
# Files created:
# users.csv
# products.csv
# events.csv
# orders.csv

import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

random.seed(42)
np.random.seed(42)

# -----------------------------
# SETTINGS
# -----------------------------
NUM_USERS = 1500
NUM_PRODUCTS = 120
NUM_EVENTS = 5000

# -----------------------------
# MASTER DATA
# -----------------------------
cities = [
    "Bangalore","Mumbai","Delhi","Chennai","Hyderabad",
    "Pune","Kolkata","Ahmedabad","Jaipur","Kochi"
]

devices = ["Mobile","Desktop","Tablet"]
channels = ["Organic","Google Ads","Instagram","LinkedIn","Referral"]

categories = {
    "Electronics": ["Earbuds","Smart Watch","Speaker","Laptop Bag","Keyboard"],
    "Fashion": ["Running Shoes","Backpack","T-Shirt","Jeans","Jacket"],
    "Home": ["Blender","Vacuum Cleaner","Lamp","Chair","Cookware"],
    "Beauty": ["Hair Dryer","Face Wash","Perfume","Trimmer","Shampoo"],
    "Books": ["Productivity Book","Finance Book","SQL Book","Python Book"]
}

# -----------------------------
# USERS TABLE
# -----------------------------
users = []

start_signup = datetime(2025,1,1)

for user_id in range(1, NUM_USERS + 1):
    signup_date = start_signup + timedelta(days=random.randint(0,180))
    
    users.append([
        user_id,
        signup_date.date(),
        random.choice(cities),
        random.choice(devices),
        random.choice(channels)
    ])

users_df = pd.DataFrame(users, columns=[
    "user_id","signup_date","city","device_type","acquisition_channel"
])

# -----------------------------
# PRODUCTS TABLE
# -----------------------------
products = []
product_id = 1001

for cat, names in categories.items():
    for i in range(NUM_PRODUCTS // len(categories)):
        name = random.choice(names) + " " + str(i+1)
        price = random.randint(499, 9999)
        rating = round(random.uniform(3.5,5.0),1)

        products.append([
            product_id,
            cat,
            name,
            price,
            rating
        ])
        product_id += 1

products_df = pd.DataFrame(products, columns=[
    "product_id","category","product_name","price","rating"
])

product_ids = products_df["product_id"].tolist()

# -----------------------------
# EVENTS + ORDERS TABLE
# -----------------------------
events = []
orders = []

event_id = 1
order_id = 5001

base_date = datetime(2025,7,1)

for _ in range(NUM_EVENTS):
    
    user = random.randint(1, NUM_USERS)
    product = random.choice(product_ids)
    
    session_time = base_date + timedelta(
        days=random.randint(0,90),
        hours=random.randint(0,23),
        minutes=random.randint(0,59)
    )
    
    # Always visit
    events.append([event_id,user,"visit",None,session_time])
    event_id += 1
    
    # 85% product view
    if random.random() < 0.85:
        t = session_time + timedelta(minutes=random.randint(1,5))
        events.append([event_id,user,"product_view",product,t])
        event_id += 1
        
        # 55% add to cart
        if random.random() < 0.55:
            t = t + timedelta(minutes=random.randint(1,4))
            events.append([event_id,user,"add_to_cart",product,t])
            event_id += 1
            
            # 65% checkout
            if random.random() < 0.65:
                t = t + timedelta(minutes=random.randint(1,3))
                events.append([event_id,user,"checkout",product,t])
                event_id += 1
                
                # 70% purchase
                if random.random() < 0.70:
                    t = t + timedelta(minutes=random.randint(1,3))
                    events.append([event_id,user,"purchase",product,t])
                    event_id += 1
                    
                    # create order
                    qty = random.randint(1,3)
                    price = int(products_df.loc[
                        products_df["product_id"] == product, "price"
                    ].values[0])
                    
                    revenue = qty * price
                    
                    orders.append([
                        order_id,
                        user,
                        product,
                        qty,
                        t.date(),
                        revenue
                    ])
                    order_id += 1

# -----------------------------
# DATAFRAMES
# -----------------------------
events_df = pd.DataFrame(events, columns=[
    "event_id","user_id","event_type","product_id","event_time"
])

orders_df = pd.DataFrame(orders, columns=[
    "order_id","user_id","product_id","quantity","order_date","revenue"
])

# -----------------------------
# EXPORT CSV
# -----------------------------
users_df.to_csv("users.csv", index=False)
products_df.to_csv("products.csv", index=False)
events_df.to_csv("events.csv", index=False)
orders_df.to_csv("orders.csv", index=False)

print("Files created successfully:")
print("users.csv")
print("products.csv")
print("events.csv")
print("orders.csv")

print("\nRows:")
print("Users:", len(users_df))
print("Products:", len(products_df))
print("Events:", len(events_df))
print("Orders:", len(orders_df))
```
